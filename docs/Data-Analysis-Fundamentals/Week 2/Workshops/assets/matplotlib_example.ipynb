### Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Read in Data

```python
sharks = pd.read_csv("cleaner_sharks.csv", index_col=0)
```

```python
sharks.info()
```

### Matplotlib

```python
type_counts = sharks['type'].value_counts() # Gets the values as numeric data
```

```python
plt.bar(type_counts.index, type_counts.values) # Create a bar chart with the values from type counts

# Set the figure size to be reasonable
plt.figure(figsize=(10, 6))

# Matplotlib uses multiple calls into plt
plt.title('Distribution of Incidents by Type')
plt.xlabel('Type of Incident')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt
```
