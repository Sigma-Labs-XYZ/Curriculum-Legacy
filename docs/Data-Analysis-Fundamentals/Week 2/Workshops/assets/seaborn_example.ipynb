```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

### Read in data

```python
sharks = pd.read_csv("./cleaner_sharks.csv", index_col=0)
```

```python
sharks.info()
```

### Prepare data

```python
# Get the indices of the countries with the top 10 number of incidents
top_countries = sharks['country'].value_counts().head(10).index

# Filter the data to only include the top 10 countries
filtered_sharks = sharks[sharks["country"].isin(top_countries)]
```

```python
# Create a reasonable sized plot using plt
plt.figure(figsize=(10, 6))

# Use seaborn to create a countplot
sns.countplot(data=filtered_sharks, x='country', palette='viridis', order=top_countries)

# Configure it like matplotlib
plt.title('Number of Shark Incidents by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=70)
plt.show()
```
