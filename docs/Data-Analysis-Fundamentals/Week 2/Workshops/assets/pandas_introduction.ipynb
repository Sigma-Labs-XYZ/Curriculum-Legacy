### Cleaning dataframes

```python
sharks.drop("pdf", axis=1)  # Drop columns that are not of interest

# axis=1 means "drop labels along the columns, not the rows"
```

**By default, Pandas always edits a _copy_. If you want to keep your changed version, you need to store it somewhere.**

```python
# Changes must be overwritten

sharks = sharks.drop("pdf", axis=1)
```

```python
# Instead of dropping, why not just select the columns you want?

sharks = sharks[["Type", "Country", "Species ", "Injury", "Date", "Sex"]]
```

```python
# Rename columns by dict

sharks = sharks.rename(columns={"Species " : "species", "Injury": "injury"})
```

```python
# Or reassign them

sharks.columns = sharks.columns.str.lower()

# .str. on a Pandas column/row (a 'Series') or an index (row labels or column labels)
# gives access to the string methods like .replace(), .lower(), or .index().
# The method is applied to every value, and the results returned.
```

```python
# Case-fold a column and remove trailing spaces

sharks["injury"] = sharks["injury"].str.lower().str.strip()  # Note, again, the use of chaining
```

```python
# Replace whole values in a column

sharks["injury"] = sharks["injury"].astype(str)
```

```python
sharks["injury"]
```

```python
# Apply a lambda to a dataframe

sharks['fatal'] = sharks["injury"].map(lambda row: "fatal" in row)
```

```python
# Change a column's type
sharks["fatal"] = sharks["fatal"].astype(bool)
```

```python
sharks["fatal"].value_counts()
```

**Don't iterate through rows in a dataframe; it's slow and inefficient. Use dataframe methods instead.**

*`.iterrows()` is **always** bad code.*

```python
# Change a column's type

sharks["type"].value_counts()
```

```python
sharks["sex"] = sharks["sex"].str.strip().replace({"lli":None, "M x 2": "M", "N": None, ".": None})
```

#### Handling missing values

```python
# Identify missing (NA) values

sharks.isna().sum()
```

```python
# Drop rows with missing values

sharks = sharks.dropna(subset=["type", "country"])
```
