```python
import pandas as pd
import altair as alt
alt.data_transformers.enable("vegafusion")
```

### Read in data

```python
sharks = pd.read_csv("./cleaner_sharks.csv", index_col=0)
```

```python
sharks.info()
```

```python
sharks["fatal"].value_counts()
```

### Create a chart

```python
alt.Chart(sharks).mark_bar().encode(
    x=alt.X('type:N', title='Type of Incident', sort='-y'),
    y=alt.Y('count()', title='Count'),
    color=alt.Color('type:N', legend=None)
).properties(
    title='Distribution of Incidents by Type',
    width=800,  # Set the width
    height=400  # Set the height
)
```

```python
alt.Chart(sharks).mark_bar().encode(
    x=alt.X("type:O", title="Type of Attack", sort="-y"),
    y=alt.Y('count():Q', title='Number of Incidents'),
    color=alt.Color("fatal:O", title="Was fatal?")
).properties(
    width=800,
    height=400,
    title='Shark Incidents Over Time by Type'
)
```

```python
top_countries = sharks['country'].value_counts().head(10).index

# Filter the data to only include the top 10 countries
filtered_sharks = sharks[sharks["country"].isin(top_countries)]

alt.Chart(filtered_sharks).mark_bar().encode(
    x=alt.X("country", sort='-y'),
    y=alt.Y("count()"),
    color="sex"
).properties(
    width=800,
    height=400
)
```

```python
sharks["fatal"].value_counts()  # We will be grouping by fatal or not

alt.Chart(sharks).mark_bar().encode(
    # alt.Y is optional, but adds more functionality
    x=alt.X("type:O", title="Type of Attack", sort="-y"),
    y=alt.Y('count():Q', title='Number of Incidents'),  # same as above
    # Color acts as a 'group by' so to speak
    color=alt.Color("fatal:O", title="Was fatal?"),
    tooltip=[
        alt.Tooltip('type:O', title='Type of Attack'),
        alt.Tooltip('fatal:O', title='Was Fatal?'),
        alt.Tooltip('count():Q', title='Number of Incidents')
    ]
).properties(  # This is how to format charts, different from matplotlib and seaborn
    width=800,
    height=400,
    title='Shark Incidents Over Time by Type'
)
```
